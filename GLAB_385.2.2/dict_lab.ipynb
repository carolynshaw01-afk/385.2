## Step 1: Strings

###  1. What is a String?
A String is a collection of characters used to represent text.
**A string is the TEXT/data itself.**
**The quotation marks are **not the string**.  
**The quotation marks tell Python where the string begins and ends.**

Example:

`'Hello'`

- `Hello` = the string value (the text/data)
- `' '` = quotation marks telling Python that `Hello` is text

### Step 5: Creating a String with Triple Single Quotes

#### What is Different About `str3`?

`str3` uses **three single quotation marks** at the beginning and three single quotation marks at the end.

`'''This is a string in Python'''`

This is another valid way to create a string in Python.

#### Why Use Triple Quotes?

Triple quotes are useful when a string needs to span multiple lines.

They allow text to continue onto additional lines while remaining one string.

**Example:**




    message = '''Hello Carolyn,
    Today we are learning Python.  
    This is still one string.'''

#### Memory Rule

- Single `' '` or double `" "` quotes → ordinary strings
- Triple `''' '''` or `""" """` quotes → especially useful for multi-line strings

**What I Expect:** Python will still print the same string value.

**Slide Reference:** Page 9

### Step 6: Code [Next Below]
### 🧠 Key Takeaways So Far

- `str3` is just a **variable name**.
- `=` is the **assignment operator** → assign/store the value on the right in the variable on the left.
- A **string is the text/data itself**.
- Quotation marks tell Python **this is text** and where the string begins and ends.
- Single `' '` and double `" "` quotes can both define strings.
- Triple quotes can be used for **multi-line strings**.
- `print(str3)` means → display the value stored in `str3`.
- **Define it first → use it second.** Python must know what a variable is before later code can use it.
#### Why Does Python Have Different Types of Quotes?

**Different quotation styles make it easier to create strings that contain quotation marks or span multiple lines.**

- Double quotes `" "` are convenient when the text contains an apostrophe: `"I'm learning Python."`
- Single quotes `' '` are convenient when the text contains double quotes: `'The word "Python" is quoted.'`
- Triple quotes `''' '''` or `""" """` are useful for strings that span multiple lines.

**Memory Rule:** Choose quotation marks that make the string easiest to write and read.



```python
# Cell Type: Code

str3 = '''This is a string in Python'''
print(str3)
# Remember:  str3 is the variable name. The triple quotes tell Python this is a string. print(str3) displays the string value.
# The quotation marks tell Python where the string begins and ends. They are not part of the string value.
```

## Exercise 1: List Manipulation 
Part 1:  Defining a List  

Create a list named `tools` containing Laptop, Monitor, Keyboard, and Mouse.

**Class Connection:** This is the same concept as creating the `cars` list in class.

```python
tools = ["Laptop", "Monitor", "Keyboard", "Mouse"] 
```

**Cell type:  Markdown

### Part 2: Indexing and Slicing

Access elements in the `tools` list using their index positions and use slicing to access part of the list.

**Class Connection:** Dylan demonstrated indexing and slicing using positions that start at `0`.

#### Understanding Index Numbers

Python lists can be accessed from the beginning or the end.

**Positive Indexing:** Counts from the left and starts at `0`.

`Laptop = 0`  
`Monitor = 1`  
`Keyboard = 2`  
`Mouse = 3`

**Negative Indexing:** Counts from the right and starts at `-1`.

`Laptop = -4`  
`Monitor = -3`  
`Keyboard = -2`  
`Mouse = -1`

**Memory Rule:**
- `0` = first item
- `-1` = last item
- Positive numbers count from the left.
- Negative numbers count from the right.

```python
# --- 2. Indexing and Slicing ---

print(tools[0])     # Laptop
print(tools[-1])    # Mouse

sub_list = tools[1:3]
print(sub_list)     # ['Monitor', 'Keyboard']
```

```python
# Cell type:  Code
# --- 3. Modifying Elements ---

tools[2] = "Webcam"
print(tools)
```
